```python
import math
from collections import Counter

import matplotlib.pyplot as plt
import torch
from torch import nn
from torch.utils import data

torch.manual_seed(42)

epoch = 300
batch_size = 64
num_steps = 20
lr = 0.1
num_examples = 600

data_path = "../data/fra-eng/fra.txt"
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("device:", device)
```

```python
def data_read(path: str) -> str:
    with open(path, "r", encoding="utf-8") as file:
        return file.read()


def preprocess_nmt(text: str) -> str:
    text = (
        text
        .replace("\u202f", " ")
        .replace("\xa0", " ")
        .lower()
    )

    result = []

    for i, char in enumerate(text):
        if (
            char in ",.!?"
            and i > 0
            and text[i - 1] != " "
        ):
            result.append(" ")

        result.append(char)

    return "".join(result)


def tokenize_nmt(text: str, max_text: int = 600):
    source = []
    target = []

    for line in text.splitlines():
        parts = line.split("\t")

        if len(parts) < 2:
            continue

        source.append(parts[0].split())
        target.append(parts[1].split())

        if max_text is not None and len(source) >= max_text:
            break

    return source, target
```

```python
def vocab_maker(tokens, min_freq: int = 2):
    counter = Counter()

    for line in tokens:
        counter.update(line)

    token_freq = sorted(
        counter.items(),
        key=lambda item: (-item[1], item[0])
    )

    idx_to_token = [
        "<unk>",
        "<pad>",
        "<bos>",
        "<eos>",
    ]

    for token, freq in token_freq:
        if freq < min_freq:
            continue

        if token not in idx_to_token:
            idx_to_token.append(token)

    token_to_idx = {
        token: idx
        for idx, token in enumerate(idx_to_token)
    }

    return {
        "idx_to_token": idx_to_token,
        "token_to_idx": token_to_idx,
    }


def truncate_pad(line, step_num: int, pad_idx: int):
    if len(line) > step_num:
        return line[:step_num]

    return line + [pad_idx] * (step_num - len(line))


def build_array_nmt(lines, vocab, step_num: int):
    token_to_idx = vocab["token_to_idx"]

    unk_idx = token_to_idx["<unk>"]
    pad_idx = token_to_idx["<pad>"]
    eos_idx = token_to_idx["<eos>"]

    array = []

    for line in lines:
        line_idx = [
            token_to_idx.get(token, unk_idx)
            for token in line
        ]

        line_idx.append(eos_idx)
        line_idx = truncate_pad(
            line_idx,
            step_num,
            pad_idx,
        )

        array.append(line_idx)

    array = torch.tensor(array, dtype=torch.long)
    valid_len = (array != pad_idx).sum(dim=1)

    return array, valid_len


def load_data_nmt(
    batch_size: int,
    step_num: int,
    max_examples: int = 600,
):
    raw_text = data_read(data_path)
    text = preprocess_nmt(raw_text)
    source, target = tokenize_nmt(text, max_examples)

    src_vocab = vocab_maker(source, min_freq=2)
    tgt_vocab = vocab_maker(target, min_freq=2)

    src_array, src_valid_len = build_array_nmt(
        source,
        src_vocab,
        step_num,
    )

    tgt_array, tgt_valid_len = build_array_nmt(
        target,
        tgt_vocab,
        step_num,
    )

    dataset = data.TensorDataset(
        src_array,
        src_valid_len,
        tgt_array,
        tgt_valid_len,
    )

    train_iter = data.DataLoader(
        dataset,
        batch_size=batch_size,
        shuffle=True,
    )

    return train_iter, src_vocab, tgt_vocab
```

```python
train_iter, src_vocab, tgt_vocab = load_data_nmt(
    batch_size,
    num_steps,
    num_examples,
)

src_pad_idx = src_vocab["token_to_idx"]["<pad>"]
tgt_pad_idx = tgt_vocab["token_to_idx"]["<pad>"]
tgt_bos_idx = tgt_vocab["token_to_idx"]["<bos>"]
tgt_eos_idx = tgt_vocab["token_to_idx"]["<eos>"]

X, X_valid_len, Y, Y_valid_len = next(iter(train_iter))

print("X shape:", X.shape)
print("X_valid_len shape:", X_valid_len.shape)
print("Y shape:", Y.shape)
print("Y_valid_len shape:", Y_valid_len.shape)
print("src vocab size:", len(src_vocab["idx_to_token"]))
print("tgt vocab size:", len(tgt_vocab["idx_to_token"]))
```

```python
def mask_softmax(
    scores: torch.Tensor,
    valid_len: torch.Tensor,
) -> torch.Tensor:
    # scores:    (batch, query_num, key_num)
    # valid_len: (batch,)

    key_num = scores.shape[-1]

    positions = torch.arange(
        key_num,
        device=scores.device,
    ).reshape(1, 1, -1)

    mask = positions >= valid_len.reshape(-1, 1, 1)
    scores = scores.masked_fill(mask, -1e6)

    return torch.softmax(scores, dim=-1)
```

加了attention的区别就是，之前是采用把输出的state改造一下与y一块合并，现在是将y拆成一个词一个词，然和和注意力机制产出的context合并作为输入

```python
class attention_module(nn.Module):
    def __init__(self,query_size,key_size,attention_hidden):
        super().__init__()
        self.w_q=nn.Linear(query_size,attention_hidden)
        self.w_k=nn.Linear(key_size,attention_hidden)
        self.w_v=nn.Linear(attention_hidden,1)
    def forward(self,query,key,value,src_valid_len): #value batch step encoder_output_size
        query=self.w_q(query) #batch query_num attention_hidden
        key=self.w_k(key) #batch step(key_num) encoder_output_size ->batch key_num attention_hidden
        features = torch.tanh(
            query.unsqueeze(2)
            + key.unsqueeze(1)
        )
        scores=self.w_v(features).squeeze(-1) #batch query_num key_num
        attention_weight=mask_softmax(scores,src_valid_len)
        context=torch.bmm(attention_weight,value) #batch query_num encoder_output_size
        return context




class encoder(nn.Module):
    def __init__(self,input_num,embedding_num,hidden_size,pad_idx):
        super().__init__()
        self.input_num=input_num
        self.embedding_num=embedding_num
        self.hidden_size=hidden_size
        self.pad_idx=pad_idx
        self.embedding=nn.Embedding(input_num,embedding_num,padding_idx=pad_idx)
        self.gru=nn.GRU(embedding_num,hidden_size,bidirectional=True)
    def forward(self,x):
        x=self.embedding(x)
        x=x.permute(1,0,2)
        output,state=self.gru(x)
        return output,state
class decoder(nn.Module):
    def __init__(self,input_num,embedding_num,hidden_size,pad_idx,encoder_output_size):
        super().__init__()
        self.input_num=input_num
        self.embedding_num=embedding_num
        self.hidden_size=hidden_size
        self.pad_idx=pad_idx
        self.attention_module=attention_module(hidden_size,encoder_output_size,256)
        self.embedding=nn.Embedding(input_num,embedding_num,padding_idx=pad_idx)
        self.gru=nn.GRU(embedding_num+encoder_output_size,hidden_size)
        self.linear=nn.Linear(hidden_size,input_num)
    def forward(self,y,state,encoder_output,src_valid_len): 
        y=self.embedding(y)
        y=y.permute(1,0,2)
        outputs = []
        for i in y: #i batch,embedding
            query=state[-1].unsqueeze(1) #batch 1 hidden
            #encoder_output  batch,step,encoder_output_size
            context=self.attention_module(query,encoder_output,encoder_output,src_valid_len) #batch 1 encoder_output_size
            i=i.unsqueeze(1) #batch 1 embedding
            gru_input=torch.cat((i,context),dim=2)
            gru_input=gru_input.permute(1,0,2)
            output1,state=self.gru(gru_input,state)
            outputs.append(output1)
        outputs=torch.cat(outputs,dim=0) #step batch hidden
        outputs=outputs.reshape(-1,self.hidden_size)
        outputs=self.linear(outputs)
        return outputs,state
        



class seq2seq(nn.Module):
    def __init__(self,encoder,decoder,bos_idx):
        super().__init__()
        self.bos_idx=bos_idx
        self.encoder=encoder
        self.decoder=decoder
        self.linear=nn.Linear(encoder.hidden_size*2,decoder.hidden_size)
    def forward(self,x,x_len,y):
        output,state=self.encoder(x)
        output=output.permute(1,0,2) #batch step encoder_hidden*2
        state1=state[0]
        state2=state[1]
        state=torch.cat((state1,state2),dim=1) 
        state=self.linear(state)
        state=torch.tanh(state)
        decoder_state=state.unsqueeze(0).repeat(self.decoder.gru.num_layers,1,1) 
        bos=torch.full(size=(y.shape[0],1),fill_value=self.bos_idx,dtype=torch.long,device=y.device)
        y=torch.cat((bos,y[:,:-1]),dim=1)
        output,state=self.decoder(y,decoder_state,output,x_len)
        return output,state

myencoder=encoder(len(src_vocab['token_to_idx']),128,256,src_vocab['token_to_idx']['<pad>'])
mydecoder=decoder(len(tgt_vocab['token_to_idx']),256,512,tgt_vocab['token_to_idx']['<pad>'],512)
net=seq2seq(myencoder,mydecoder,tgt_vocab['token_to_idx']['<bos>']).to(device)

trainer=torch.optim.SGD(net.parameters(),lr=lr)

loss = nn.CrossEntropyLoss(
    ignore_index=tgt_pad_idx
)


```

```python
for i in range(epoch):
    net.train()
    train_loss = 0.0
    train_token_count = 0
    for x,x_len,y,y_len in train_iter:
        x=x.to(device)
        y=y.to(device)
        x_len=x_len.to(device)
        trainer.zero_grad()
        logit,state=net(x,x_len,y)
        y=y.T.reshape(-1)
        l=loss(logit,y)
        l.backward()
        nn.utils.clip_grad_norm_(
            net.parameters(),
            max_norm=1.0,
        )
        trainer.step()
        valid_token_num = y_len.sum().item()
        train_loss += l.item() * valid_token_num
        train_token_count += valid_token_num

    average_loss = train_loss / train_token_count
    print(f"epoch{i},loss:{average_loss:.5f}")

```

```python
def predict_seq2seq(
    text,
    net,
    src_vocab,
    tgt_vocab,
    num_steps,
    device
):
    net.eval()

    text = preprocess_nmt(text)
    tokens = text.split()

    src_token_to_idx = src_vocab["token_to_idx"]
    tgt_token_to_idx = tgt_vocab["token_to_idx"]

    src_unk_idx = src_token_to_idx["<unk>"]
    src_pad_idx = src_token_to_idx["<pad>"]
    src_eos_idx = src_token_to_idx["<eos>"]

    bos_idx = tgt_token_to_idx["<bos>"]
    eos_idx = tgt_token_to_idx["<eos>"]

    x = [
        src_token_to_idx.get(token, src_unk_idx)
        for token in tokens
    ]

    x.append(src_eos_idx)

    x_len = torch.tensor(
        [min(len(x), num_steps)],
        dtype=torch.long,
        device=device
    )

    x = truncate_pad(
        x,
        num_steps,
        src_pad_idx
    )

    x = torch.tensor(
        x,
        dtype=torch.long,
        device=device
    ).reshape(1, -1)

    result_idx = []

    with torch.no_grad():
        encoder_output, encoder_state = net.encoder(x)

        encoder_output = encoder_output.permute(1, 0, 2)

        forward_state = encoder_state[0]
        backward_state = encoder_state[1]

        decoder_state = torch.cat(
            (forward_state, backward_state),
            dim=1
        )

        decoder_state = torch.tanh(
            net.linear(decoder_state)
        )

        decoder_state = decoder_state.unsqueeze(0).repeat(
            net.decoder.gru.num_layers,
            1,
            1
        )

        decoder_x = torch.tensor(
            [[bos_idx]],
            dtype=torch.long,
            device=device
        )

        for _ in range(num_steps):
            logit, decoder_state = net.decoder(
                decoder_x,
                decoder_state,
                encoder_output,
                x_len
            )

            predict_idx = logit.argmax(dim=1).item()

            if predict_idx == eos_idx:
                break

            result_idx.append(predict_idx)

            decoder_x = torch.tensor(
                [[predict_idx]],
                dtype=torch.long,
                device=device
            )

    result = [
        tgt_vocab["idx_to_token"][idx]
        for idx in result_idx
    ]

    return " ".join(result)


predict_seq2seq(
    "go .",
    net,
    src_vocab,
    tgt_vocab,
    num_steps,
    device
)
```
