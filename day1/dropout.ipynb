```python
import torch
from torch import nn
from torch.utils import data
from torchvision import datasets, transforms
import matplotlib.pyplot as plt
```

```python
def data_maker():
    trans=transforms.ToTensor()
    train_dataset=datasets.FashionMNIST(root='../data',train=True,transform=trans,download=True)
    test_dataset=datasets.FashionMNIST(root='../data',train=False,transform=trans,download=True)
    return train_dataset,test_dataset
```

```python
def data_loader(data_array,batch_size,shuffle=True):
    return data.DataLoader(data_array,batch_size,shuffle=shuffle)

```

```python
def init(m):
    if type(m)==nn.Linear:
        nn.init.normal_(m.weight,mean=0,std=0.01)
        nn.init.zeros_(m.bias)
        
```

```python
net=nn.Sequential(nn.Flatten(start_dim=1,end_dim=-1),nn.Linear(784,256),nn.ReLU(),nn.Dropout(0.2),nn.Linear(256,256),nn.ReLU(),nn.Dropout(0.5),nn.Linear(256,10))
net.apply(init)

loss=nn.CrossEntropyLoss()

weightdecay=0
lr=0.1

decay_params=[]
no_decay_params=[]
for name,param in net.named_parameters():
    if name.endswith("bias"):
        no_decay_params.append(param)
    else:
        decay_params.append(param)

trainer=torch.optim.SGD(
    [
        {
            "params":decay_params,
            "weight_decay":weightdecay
        },
        {
            "params":no_decay_params
        }
    ],
    lr=lr

)
batch_size=256
train_data,test_data=data_maker()

train_iter=data_loader(train_data,batch_size,shuffle=True)
test_iter=data_loader(test_data,batch_size,shuffle=False)

epoch=20

for i in range(epoch):
    net.train()
    for x,y in train_iter:
        trainer.zero_grad()
        l=loss(net(x),y)
        l.backward()
        trainer.step()
    net.eval()
    total=0
    with torch.no_grad():
        for x,y in test_iter:
            logit=net(x)
            num=(logit.argmax(dim=1)==y).sum().item()
            total+=num
            l=loss(logit,y)
    print(f"{total/len(test_data)}")

```
