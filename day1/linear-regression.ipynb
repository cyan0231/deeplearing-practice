```python
import torch
from torch.utils import data
from torch import nn
```

```python
def data_maker(w,b):
    x=torch.normal(0,1,(1000,2))
    y=torch.matmul(x,w)+b
    y+=torch.normal(0,0.01,size=y.shape)
    return x,y
```

```python
def data_loader(data_array,batch_size):
    dataset=data.TensorDataset(*data_array)
    return data.DataLoader(dataset,batch_size,shuffle=True)
```

```python
net = nn.Sequential(nn.Linear(2, 1)) # model

with torch.no_grad():  # init
    net[0].weight.normal_(0, 0.01)
    net[0].bias.fill_(0)

loss_fn = nn.MSELoss() # loss

trainer = torch.optim.SGD(   #optim
    net.parameters(),
    lr=0.01
)
```

```python
true_w = torch.tensor([2, -3.4]).reshape(2,1)
true_b = 4.2
epoch=10
batch_size=10
features,labels=data_maker(true_w,true_b)
print(labels.size())
data_iter=data_loader((features,labels),batch_size)

```

```python

for i in range(epoch):
    for x,y in data_iter:
        l=loss_fn(net(x),y)
        trainer.zero_grad()
        l.backward()
        trainer.step()
    with torch.no_grad():

        l=loss_fn(net(features),labels)
        print(l.item())
```
