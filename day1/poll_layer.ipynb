```python
import torch
from torch import nn
```

```python
X = torch.arange(16, dtype=torch.float32).reshape((1, 1, 4, 4))
print(X.size())
X = torch.cat((X, X + 1), 1)
print(X.size())
net=nn.MaxPool2d(kernel_size=(3,3),padding=(1,1),stride=(2,2))#要求输入的x维度是（batch_size,channel,h,w）
print(net(X))
net1=nn.AvgPool2d(kernel_size=(3,3),padding=(1,1),stride=(2,3))
print(net1(X))
```

```python
class minpool(nn.Module):
    def __init__(self,kernel_size,stride=None,padding=0):
        super().__init__()
        self.net=nn.MaxPool2d(kernel_size=kernel_size,stride=stride,padding=padding)
    def forward(self,x):
        return -self.net(-x)
```

```python
net2=minpool(kernel_size=(2,2))
x = torch.tensor([[[[
    1.0, 4.0,
    3.0, 2.0
]]]]).reshape(1,1,2,2)
print(net2(x))
net2(x).size()
```
