```python
class mynet(nn.Module):
    def __init__(self):
        super().__init__()
        self.net11=nn.Sequential()
        self.net11.add_module("1",nn.Linear(20,20))
    def forward(self,x):
        for layer in self.children():
            x=layer(x)
        return x
x=torch.randn((20,20))
net=mynet()
y=net(x)
#print(net.net[0].weight)
torch.save(net.state_dict(),"net_data")
#print(net.state_dict().values())
net1=mynet()
state=torch.load('net_data')
#print(state)
net1.load_state_dict(state)
print(state.keys())
print(net)


```

```python
def mycompare(model1,model2):
    struct1=[
        (name,type(module),module.extra_repr())
        for name,module in model1.named_modules()
    ]
    struct2=[
        (name,type(module),module.extra_repr())
        for name,module in model2.named_modules()
    ]
    same_structure=(struct1==struct2)
    state1=model1.state_dict()
    state2=model2.state_dict()
    same_key=state1.keys()==state2.keys()

    value1=[]
    value2=[]
    for i,j in zip(state1.keys(),state2.keys()):
        value1.append(state1[i].all())
        value2.append(state2[j].all())
    same_value=value1==value2
    
    return same_structure,same_key,same_value
```

```python
a1,a2,a3=mycompare(net,net1)
print(a1,a2,a3)
```
