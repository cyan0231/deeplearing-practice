```python
import torch
from torch import nn
```

```python
conv_arch = ((1, 64), (1, 128), (2, 256), (2, 512), (2, 512))
```

```python
class vgg_block(nn.Module):
    def __init__(self,num_convs,in_channel,out_channel):
        super().__init__()
        self.net=nn.Sequential()
        for i in range(num_convs):
            self.net.append(nn.Conv2d(in_channel,out_channel,kernel_size=3,padding=1))
            self.net.append(nn.ReLU())
            in_channel=out_channel
        self.net.append(nn.MaxPool2d(kernel_size=2,stride=2))
    def forward(self,x):
        return self.net(x)

```

```python
class vgg(nn.Module):
    def __init__(self,conv_arch):
        super().__init__()
        self.net=nn.Sequential()
        in_channel=1
        for num,out_channel in conv_arch:
            self.net.append(vgg_block(num,in_channel,out_channel))
            in_channel=out_channel
        self.net.append(nn.Flatten())
        self.net.append(nn.Linear(out_channel*7*7,4096))
        self.net.append(nn.ReLU())
        self.net.append(nn.Dropout(0.5))
        self.net.append(nn.Linear(4096,4096))
        self.net.append(nn.ReLU())
        self.net.append(nn.Dropout(0,5))
        self.net.append(nn.Linear(4096,10))
    def forward(self, x):
        return self.net(x)
```

```python
net=vgg(conv_arch)
print(net)
```

```python
def vgg_block1(num_convs, in_channels, out_channels):
    layers = []
    for _ in range(num_convs):
        layers.append(nn.Conv2d(in_channels, out_channels,
                                kernel_size=3, padding=1))
        layers.append(nn.ReLU())
        in_channels = out_channels
    layers.append(nn.MaxPool2d(kernel_size=2,stride=2))
    return nn.Sequential(*layers)
```

```python
def vgg1(conv_arch):
    conv_blks = []
    in_channels = 1
    # 卷积层部分
    for (num_convs, out_channels) in conv_arch:
        conv_blks.append(vgg_block1(num_convs, in_channels, out_channels))
        in_channels = out_channels

    return nn.Sequential(
        *conv_blks, nn.Flatten(),
        # 全连接层部分
        nn.Linear(out_channels * 7 * 7, 4096), nn.ReLU(), nn.Dropout(0.5),
        nn.Linear(4096, 4096), nn.ReLU(), nn.Dropout(0.5),
        nn.Linear(4096, 10))

net = vgg1(conv_arch)
print(net)
```
